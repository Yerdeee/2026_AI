```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits

# 데이터셋 불러오기
digits = load_digits()

print("데이터 개수:", len(digits.images))
print("이미지 데이터 shape:", digits.images.shape)
print("정답 데이터 shape:", digits.target.shape)
print("클래스:", digits.target_names)
```

```python
index = 0

image = digits.images[index]
label = digits.target[index]

print("이미지 shape:", image.shape)
print("정답 label:", label)
print("픽셀 최솟값:", image.min())
print("픽셀 최댓값:", image.max())

plt.imshow(image, cmap="gray")
plt.title(f"answer: {label}")
plt.axis("off")
plt.show()
```

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 5))

for ax in axes.flatten():
    index = np.random.randint(0, len(digits.images))

    image = digits.images[index]
    label = digits.target[index]

    ax.imshow(image, cmap="gray")
    ax.set_title(f"label: {label}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 5))

for number, ax in enumerate(axes.flatten()):
    index = np.where(digits.target == number)[0][0]

    ax.imshow(digits.images[index], cmap="gray")
    ax.set_title(f"number: {number}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```
